## import Model

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")  # or your custom model, e.g. "model101.pt"
```

## 1. Detection from Image

```python

img_path = "assets/numl1.jpg"
results = model.predict(source=img_path, save=True, show=True, project="detections", conf=0.3)  # adjust confidence if needed
```

## 2. Detection From Video

```python
video_path = "assets/tour.mp4"

results = model.predict(source=video_path, save=True, show=True, conf=0.3)
```

## 3. Live Detection From Webcam

```python
results = model.predict(source=0, show=True, conf=0.3)
```

# Understanding Parameters


- source : path of image, video and for webcam source=0
- show : Show Frame 
- save: save the detected file
- conf : percentage e.g 0.5 ( will detect only the object having confidence 50 or >)
- project : Output folder
